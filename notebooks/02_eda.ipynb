# Phase 4: Exploratory Data Analysis
Answering key questions about CP Bot user performance.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../ml/data/cleaned_dataset.csv")
features_df = pd.read_csv("../ml/data/features.csv")

sns.set_theme(style="whitegrid")
```

### Q1: How does problem difficulty affect solve probability?

```
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x="problem_rating", hue="solved", multiple="stack", bins=20)
plt.title("Problem Rating vs Solve Probability")
plt.show()
```

### Q4: Does rating difference predict success?

```
plt.figure(figsize=(10, 6))
sns.boxplot(data=features_df, x="solved", y="rating_difference")
plt.title("Rating Difference by Solve Outcome")
plt.show()
```
